```python
import pandas as pd
df = pd.read_csv("/content/SampleData.csv")
print(df)
```

```python
print("Missing value in each column")
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
print("missing values after handling")
print(df.isnull().sum())
```

```python
print("gender values before fomatting")
print(df['Gender'].unique())
```

```python
df['Gender'] = df['Gender'].str.strip().str.upper()
print("gender values after formatting")
print(df['Gender'].unique())
```

```python
Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)
IQR = Q3 = Q1
print("Q1: ",Q1)
print("Q3: ",Q3)
print("IQR: ",IQR)
```

```python
Lower_limit = Q1 - (1.5 * IQR)
Upper_limit = Q3 + (1.5 * IQR)
print("Lower limit: ", Lower_limit)
print("Upper limit: ", Upper_limit)
```

```python
outliers = df[
    (df['Marks'] < Lower_limit)|
    (df['Marks'] > Upper_limit)
]
print("Outliers:")
print(outliers)
```

```python
df['Marks']=df['Marks'].clip(
    lower = Lower_limit,
    upper = Upper_limit
)
print("outlier handeled successfully")
```

```python
print("Final Clean Dataset:")
print(df)
```
