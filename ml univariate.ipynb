```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/drive/MyDrive/ML/SampleData.csv")

df['Gender'] = df ['Gender'].str.strip().str.upper()

print("\n Dataset after cleaning:")
print(df)

print("\n Descriptive Summary Statistics: ")
print(df.describe())

plt.hist(df['Marks'], bins=5)

plt.xlabel('Marks')
plt.ylabel('number of students')
plt.title('Distribution of Marks')
plt.show()

gender_count = df['Gender'].value_counts()
plt.bar(gender_count.index, gender_count.values)
plt.xlabel('Gender')
plt.ylabel('Number of Students')
plt.title('Number of Students by Gender')
plt.show()

plt.scatter(df['Attendance'], df['Marks'])
plt.xlabel('Attendance')
plt.ylabel('Marks')
plt.title('Attendance vs Marks')
plt.show()

sns.boxplot(data=df, x='Gender', y='Marks')
plt.xlabel('Gender')
plt.ylabel('Marks')
plt.title('Box Plot of Marks by Gender')
plt.show()

features = ['Age', 'Gender', 'Attendance']
target = 'Marks'

print("\n Features:")
print(features)
print("\n Target:")
print(target)

print("\n practical completed successfully!")
```
