```python
import pandas as pd
data = {
    'Attendance': ['High', 'High', 'Low', 'High'],
    'Marks': ['Excellent', 'Excellent', 'High', 'High'],
    'Income': ['Low', 'Medium', 'High', 'High'],
    'Sports': ['Yes', 'Yes', 'No', 'Yes'],
    'Scholarship': ['Yes', 'Yes', 'No', 'Yes'],
}
df = pd.DataFrame(data)

df.to_csv('student_scholarship.csv', index = False)

print("Training Dataset")
display(df)

df = pd.read_csv('student_scholarship.csv')

attributes = list(df.columns[:-1])

target = df.columns[-1]

hypothesis = [None] * len(attributes)

for _, row in df.iterrows():
    if row[target] == 'Yes':
        for i in range(len(attributes)):
            if hypothesis[i] is None:
                hypothesis[i] = row[attributes[i]]
            elif hypothesis[i] != row[attributes[i]]:
              hypothesis[i] = '?'

print("Hypothesis: ", hypothesis)
print("\nAttribute-wise Hypothesis:")

for i in range(len(attributes)):
  print(attributes[i], "=", hypothesis[i])
```
